# Example of a notebook

**Jupyter Notebook** is a web application that makes it possible to produce documents containing formatted text, images, videos... but also directly executable code (see http://jupyter.org/).

**Jupyter Notebook** is a client-server web application. If the notebook is run in your own environment, your machine plays the roles of both the client and the server. When using the **JupyterHub**, a remote server runs the notebook and the result is displayed on your client machine.

A **Jupyter Notebook** consists of a succession of cells of different types: 
* "markdown" cells to display text 
* "code" cells containing code that can be executed in the notebook

## Starting the server 

To start the server, simply type in a terminal or in the anaconda prompt (on Windows):
```bash
> jupyter notebook
```

or


## Markdown cells

To write formatted text, you have to use the **Markdown** language. 
This language makes it possible in particular:

* to write titles thanks to the # character:
  ```
  # Title  
  ## Subtitle
  ```

* to write bullet lists:
  ```markdown
  * first item
  * second item    
  * third item
  ```
  result:
  * first item
  * second item    
  * third item

* to write numbered lists:
  ```markdown
  1. first item
  2. second item    
  3. third item
  ```
  result:
  1. first item
  2. second item    
  3. third item

* to write text in bold
  ```
  **text in bold**
  ```
  result: 
  
  **text in bold**

* to write text in italics
  ```
  *text in italics*
  ```
  result: 
  
  *text in italics* 

Markdown cells also make it possible to write latex in mathematical mode in order to insert equations:

### The Python *kernel*

By default, when a **Jupyter Notebook** server is started, a Python interpreter (or Python *kernel*) has been launched and waits for instructions to be sent to it.

````python
a = 74
b = 10
print(a+b)
````

It is possible to write more complex code using several Python modules with graphical output:

````python
import plotly.graph_objects as go
import numpy as np

x = np.linspace(0, 1, 1000)

f = np.linspace(1, 16, 31)

fig = go.Figure()

# Add invisible traces, one for each slider step
for i, fi in enumerate(f):
    fig.add_trace(go.Scatter(visible=False, x=x, y=np.sin(2*np.pi*fi*x), name=f"f={fi}"))

# Make first trace visible
fig.data[0].visible = True

# Create and add slider
steps = []
for i, fi in enumerate(f):
    step = dict(method="update", label = f"{fi}", args=[{"visible": [(el==i) for el in range(f.size)]}])
    steps.append(step)
    
sliders = [dict(currentvalue={'prefix': 'Frequency  f = '}, steps=steps)]

fig.update_layout(sliders=sliders, title = 'f(x) = sin(2 pi f x)', height=500)
fig.show()
````
